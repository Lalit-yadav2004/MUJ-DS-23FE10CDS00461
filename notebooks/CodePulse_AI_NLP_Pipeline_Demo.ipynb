## 1. AST-Guided Semantic Tokenization
Parse source code into syntactic tokens, call graphs, dangerous sinks, and defensive sanitizers.

```python
sample_code = """import os, sqlite3

def authenticate_user(cursor, username, password):
    query = f"SELECT * FROM users WHERE username = '{username}' AND password = '{password}'"
    cursor.execute(query)
    return cursor.fetchone()

def ping_server(host):
    os.system("ping -c 1 " + host)
"""

structure = ASTAnalyzer.analyze_file("demo_target.py", content=sample_code)
print("Detected Language:", structure.language)
print("Dangerous Sinks Found:", structure.dangerous_sinks)
print("Functions:", list(structure.functions.keys()))
```

## 2. Execute Full Multi-Agent Triage Pipeline

```python
settings = load_settings()
settings.llm.default_provider = "mock"  # Can be 'gemini', 'nvidia', or 'mock'
orchestrator = AgentOrchestrator(settings)

report = orchestrator.analyze_file("demo_target.py", code_content=sample_code)
print(f"Overall Status: {report.overall_status}")
print("Summary Statistics:")
print(json.dumps(report.summary_statistics, indent=2))
```

## 3. Inspect Synthesized Patches & 6-Check Validation Verdicts

```python
print(f"Total Patches Generated: {len(report.patches)}\n")
for p in report.patches:
    print(f"--- [{p.candidate_id}] {p.cwe_id} ({p.file_path}) ---")
    print(f"Summary: {p.patch_summary}")
    print(f"Unified Diff:\n{p.unified_diff}\n")

print("Validation Verdicts (Stage 4):")
for v in report.patch_validations:
    print(f"- {v.candidate_id}: Verdict={v.verdict} (All Checks Passed={not v.regenerate})")
    for chk, status in v.checks.items():
        print(f"    {chk}: {status}")
```
